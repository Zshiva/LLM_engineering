```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown
```

```python
load_dotenv()

google_api_key = os.getenv('GOOGLE_API_KEY')
if not google_api_key:
    print("No API key was found - please be sure to add your key to the .env file, and save the file! Or you can skip the next 2 cells if you don't want to use Gemini")
elif not google_api_key.startswith("AIz"):
    print("An API key was found, but it doesn't start AIz")
else:
    print("API key found and looks good so far!")
```

```python
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
client = OpenAI(base_url=gemini_url, api_key=google_api_key)
```

```python
tell_a_joke = [
    {"role": "user", "content" : "Tell a joke for a student on the journey to become an expert in LLM Engineering"},
]
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash-lite", 
    messages = tell_a_joke)
print(response.choices[0].message.content)
```

```python
gemini_model = "gemini-2.5-flash"
ollama_model = "llama3.2"
gemini_system = "You are a chatbot who is argumentativee;\
                you disagree with anything in the conversatrion, and you always try to prove the other person wrong."
llama_system = "You are a helpful assistant who tries to help the user with their questions, and you always try to be right in the conversatrion."
gpt_message = ["Hi there"]
llama_message = ["Hi"]
```

```python
[{"role": "system", "content": "system_message_here"},
{
    "role": "user",
    "content": "first_user_message_here"
},
{
    "role": "assistant",
    "content": "the assistant's response to the first user message here"
},
{
    "role": "user",
    "content": "the new user prompt here"
}
]
```

```python
def call_gemini():
     messages = [
        {"role": "system", "content": gemini_system}]
     for gemini, llama in zip(gpt_message, llama_message):
        messages.append({"role": "assistant", "content": gemini})
        messages.append({"role": "user", "content": llama})
     response = client.chat.completions.create(
        model=gemini_model,
        messages=messages)
     return response.choices[0].message.content
```

```python
call_gemini()
```

```python
ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama" )
```

```python
def call_llama():
    messages = [
        {"role": "system", "content": llama_system}]
    for gemini, llama_msg in zip(gpt_message, llama_message):
        messages.append({"role": "user", "content": gemini})
        messages.append({"role": "assistant", "content": llama_msg})
    messages.append({"role": "user", "content": gpt_message[-1]})
    response = ollama.chat.completions.create(
        model="llama3.2",
        messages=messages)
    return response.choices[0].message.content
```

```python
call_llama()
```

```python
gemini_messages = ["Hi whatsupp"]
llama_messages = ["Hi there! How can I assist you today?"]

display(Markdown(f"### GEMINI: \n{gemini_messages[0]}"))
display(Markdown(f"### LLAMA: \n{llama_messages[0]}"))


for i in range(15):
    gemini_reply = call_gemini()
    display(Markdown(f"### GEMINI: \n{gemini_reply}"))
    gemini_messages.append(gemini_reply)
    llama_reply = call_llama()
    display(Markdown(f"### LLAMA: \n{llama_reply}"))
    llama_messages.append(llama_reply)
    
```
