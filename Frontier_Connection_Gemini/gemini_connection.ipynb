```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown
```

```python
load_dotenv()

google_api_key = os.getenv('GOOGLE_API_KEY')
if not google_api_key:
    print("No API key was found - please be sure to add your key to the .env file, and save the file! Or you can skip the next 2 cells if you don't want to use Gemini")
elif not google_api_key.startswith("AIz"):
    print("An API key was found, but it doesn't start AIz")
else:
    print("API key found and looks good so far!")
```

```python
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
client = OpenAI(base_url=gemini_url, api_key=google_api_key)
```

```python
tell_a_joke = [
    {"role": "user", "content" : "Tell a joke for a student on the journey to become an expert in LLM Engineering"},
]
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash-lite", 
    messages = tell_a_joke)
print(response.choices[0].message.content)
```

```python
easy_puzzle = [

    { "role": "user", "content":
        "You toss 2 coins. One of them is heads. Whats the probability that the other is tails? Answer with the probability only"}
]
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash-lite", 
    messages = easy_puzzle,
    reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash", 
    messages = easy_puzzle,
    reasoning_effort="low")
display(Markdown(response.choices[0].message.content))
```

```python
hard_message = [
    {
        "role" : "user", "content":
        "Follow the instructions carefully: "
        "Take the any word you want  "
        "Remove the 3rd and 6th letters "
        "Reverse the remaining letters"
        "Convert each letter to its position in the alphabet (A=1, B=2, ..., Z=26)"
        "Add all the numbers together "
        "Show the word you choose at last"
        "Also show how you solve the word"
        "If the result is even → output 'SMART'"
        "If the result is odd → output 'GENIUS'"

    }
]
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash-lite", 
    messages = hard_message,
    reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash", 
    messages = hard_message,
    reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```
